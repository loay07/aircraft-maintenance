```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
train = pd.read_csv("../../data/processed/train_FD001_processed.csv")
```

```python
import matplotlib.pyplot as plt

def plot_sensor(engine_id, sensor):
    engine = train[train["engine_id"] == engine_id]

    plt.figure(figsize=(10, 5))

    plt.scatter(
        engine["cycle"],
        engine[sensor],
        s=15
    )

    plt.xlabel("Operating Cycle")
    plt.ylabel(sensor)
    plt.title(f"Engine {engine_id} - {sensor} Over Time")
    plt.grid()

    plt.show()
```

```python
plot_sensor(7, "sensor_1")
```

SENSOR 1:
- constant
- consistent through all engines
- no relation with RUL

```python
plot_sensor(8, "sensor_2")
```
